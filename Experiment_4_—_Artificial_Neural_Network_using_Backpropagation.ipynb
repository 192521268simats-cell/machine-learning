```python

import numpy as np

# XOR dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)

Y = np.array([
    [0],
    [1],
    [1],
    [0]
], dtype=float)

np.random.seed(42)

# Network: 2 input neurons, 4 hidden neurons, 1 output neuron
W1 = np.random.randn(2, 4)
b1 = np.zeros((1, 4))
W2 = np.random.randn(4, 1)
b2 = np.zeros((1, 1))

learning_rate = 0.5
epochs = 20000

def sigmoid(x):
    return 1 / (1 + np.exp(-np.clip(x, -500, 500)))

# Training using backpropagation
for epoch in range(epochs):
    # Forward propagation
    hidden = sigmoid(np.dot(X, W1) + b1)
    output = sigmoid(np.dot(hidden, W2) + b2)

    # Output error
    error = output - Y

    # Backpropagation
    delta_output = error * output * (1 - output)
    delta_hidden = np.dot(delta_output, W2.T) * hidden * (1 - hidden)

    # Update weights and biases
    W2 -= learning_rate * np.dot(hidden.T, delta_output) / len(X)
    b2 -= learning_rate * np.mean(delta_output, axis=0, keepdims=True)

    W1 -= learning_rate * np.dot(X.T, delta_hidden) / len(X)
    b1 -= learning_rate * np.mean(delta_hidden, axis=0, keepdims=True)

# Test the trained network
hidden = sigmoid(np.dot(X, W1) + b1)
predictions = sigmoid(np.dot(hidden, W2) + b2)

print("XOR Dataset Predictions:")
for i in range(len(X)):
    predicted_class = 1 if predictions[i][0] >= 0.5 else 0
    print(
        X[i].astype(int),
        "Target =", int(Y[i][0]),
        "Predicted =", predicted_class,
        "Output =", round(predictions[i][0], 4)
    )
```
