```python

import csv

# Create CSV file
data = [
    ["Sky", "AirTemp", "Humidity", "Wind", "Water", "Forecast", "EnjoySport"],
    ["Sunny", "Warm", "Normal", "Strong", "Warm", "Same", "Yes"],
    ["Sunny", "Warm", "High", "Strong", "Warm", "Same", "Yes"],
    ["Rainy", "Cold", "High", "Strong", "Warm", "Change", "No"],
    ["Sunny", "Warm", "High", "Strong", "Cool", "Change", "Yes"]
]

with open("training_data.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

# Read CSV file
with open("training_data.csv", "r") as file:
    reader = csv.reader(file)
    rows = list(reader)

examples = rows[1:]
n = len(rows[0]) - 1

# Initialize boundaries
S = [None] * n
G = [["?"] * n]

# Check whether a hypothesis covers an example
def covers(h, x):
    return all(h[i] == "?" or h[i] == x[i]
               for i in range(len(h)))

# Check whether h1 is at least as general as h2
def more_general(h1, h2):
    return all(a == "?" or a == b
               for a, b in zip(h1, h2))

# Process training examples
for row in examples:
    x = row[:-1]
    target = row[-1]

    if target == "Yes":
        # Remove inconsistent general hypotheses
        G = [g for g in G if covers(g, x)]

        # Generalize specific hypothesis
        for i in range(n):
            if S[i] is None:
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = "?"

    else:
        # Remove inconsistent specific hypothesis
        if covers(S, x):
            S = [None] * n

        # Specialize general hypotheses
        new_G = []
        for g in G:
            if covers(g, x):
                for i in range(n):
                    if g[i] == "?":
                        if S[i] is not None and S[i] != x[i]:
                            h = g.copy()
                            h[i] = S[i]
                            new_G.append(h)
            else:
                new_G.append(g)
        G = new_G

# Display results
print("Training data read from CSV successfully.")
print("\nSpecific Boundary (S):")
print(S)

print("\nGeneral Boundary (G):")
for h in G:
    print(h)
```
