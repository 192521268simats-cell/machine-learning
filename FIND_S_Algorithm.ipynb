```python
# FIND-S Algorithm

# Training data
# Attributes: Study Time, Attendance, Assignment, Previous Test
data = [
    ["Long", "High", "Yes", "Good", "Yes"],
    ["Short", "High", "Yes", "Good", "No"],
    ["Long", "High", "Yes", "Average", "Yes"],
    ["Long", "Low", "Yes", "Good", "Yes"],
    ["Short", "Low", "No", "Average", "No"]
]

# Initialize the most specific hypothesis
hypothesis = ["Ø", "Ø", "Ø", "Ø"]

# Apply FIND-S algorithm
for row in data:
    attributes = row[:-1]
    target = row[-1]

    # Consider only positive examples
    if target == "Yes":

        # First positive example
        if hypothesis[0] == "Ø":
            hypothesis = attributes.copy()

        else:
            # Generalize hypothesis where necessary
            for i in range(len(hypothesis)):
                if hypothesis[i] != attributes[i]:
                    hypothesis[i] = "?"

# Display final hypothesis
print("Most Specific Hypothesis:")
print(hypothesis)
```
